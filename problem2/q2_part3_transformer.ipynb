```python
import os, time, warnings, inspect
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score, accuracy_score

SEED = 42
np.random.seed(SEED)
torch.manual_seed(SEED)

df = pd.read_pickle(f'{BASE}/q2_clean_split.pkl')
test = pd.read_pickle(f'{BASE}/q2_test_clean.pkl')

MAX_TRAIN = 150_000
tr_all = df[df['split'] == 'train']
tr, _ = train_test_split(tr_all, train_size=MAX_TRAIN, stratify=tr_all['overall'], random_state=SEED)
tr = tr.copy()
va = df[df['split'] == 'val'].copy()
print(f'train: {len(tr):,} | val: {len(va):,} | test: {len(test):,}')
print('sahm emtiaz-ha dar train:', tr['overall'].value_counts(normalize=True).sort_index().round(4).tolist())
```

<p dir="rtl" style="margin:0px 0 2px 0;font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:23px;font-weight:bold;color:#5ebcff;line-height:1.7">قدم 2. توکنایزر، پاک‌سازی سبک و انتخاب طول</p>
<p dir="rtl" style="margin:2px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">مسئله:</span> باید سقف طول ورودی انتخاب می‌شد. طول بیشتر یعنی نظرهای بیشتری کامل دیده می‌شوند، ولی آموزش کندتر می‌شود.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">توضیح:</span> برخلاف TF-IDF، اینجا حروف کوچک نمی‌شوند و استاپ‌ورد هم حذف نمی‌شود. مدل روی متن خام پیش‌آموزش دیده و چیزهایی مثل «NOT» یا «!!!» برایش معنی دارند. فقط HTML و فاصله‌های اضافه پاک می‌شوند و خلاصه به 32 کلمه کوتاه می‌شود. طول‌ها روی 5,000 نمونه اندازه‌گیری شد.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">راه حل:</span> با 384 توکن 94.9 درصد نظرها کامل جا می‌شوند و با 512 توکن 97.4 درصد؛ میانه 103 توکن است. 512 فقط 2.5 درصد پوشش بیشتر می‌داد، ولی هزینه‌ی attention با مربع طول بالا می‌رود. پس 384 انتخاب شد. هشدار «873 > 512» مال همین اندازه‌گیری بدون برش است و در آموزش اثری ندارد.</p>

```python
import re, html
from transformers import AutoTokenizer

MODEL_NAME = 'distilbert/distilroberta-base'
tok = AutoTokenizer.from_pretrained(MODEL_NAME)

TAG = re.compile(r'<[^>]+>')
SPACES = re.compile(r'\s+')
def light_clean(t):
    t = html.unescape(t)
    t = TAG.sub(' ', t)
    return SPACES.sub(' ', t).strip()

def short_summary(s, n=32):
    return ' '.join(s.split()[:n])[:200]

t0 = time.time()
for d in (tr, va, test):
    d['rt_l'] = d['reviewText'].map(light_clean)
    d['sm_l'] = d['summary'].map(light_clean).map(short_summary)
print(f'paksazi: {time.time() - t0:.0f} sanie')

sample = tr.sample(5000, random_state=SEED)
enc = tok(list(sample['sm_l']), list(sample['rt_l']), truncation=False)
lens = np.array([len(x) for x in enc['input_ids']])
for L in [128, 256, 384, 512]:
    print(f'MAX_LEN={L}: {np.mean(lens <= L):.1%} nazar-ha kamel ja mishavand')
print('sadak 50/90/95/99:', np.percentile(lens, [50, 90, 95, 99]).astype(int).tolist())
print('\nnemune voroodi model:', tok.decode(enc['input_ids'][0])[:250])
```

<p dir="rtl" style="margin:0px 0 2px 0;font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:23px;font-weight:bold;color:#5ebcff;line-height:1.7">قدم 3. ساخت دیتاست‌ها</p>
<p dir="rtl" style="margin:2px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">مسئله:</span> وقتی نظری بلندتر از سقف است و باید بریده شود، خلاصه نباید بریده شود. خلاصه کوتاه است و بیشترین معنی را دارد.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">توضیح:</span> ورودی هر نمونه یک جفت (خلاصه، متن) است و <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;background:transparent;padding:0;direction:ltr;unicode-bidi:isolate">truncation='only_second'</span> فقط متن را کوتاه می‌کند. برچسب <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;background:transparent;padding:0;direction:ltr;unicode-bidi:isolate">overall - 1</span> است چون مدل کلاس‌ها را از 0 می‌شمارد. در خروجی نهایی دوباره 1 اضافه می‌شود.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">راه حل:</span> میانگین طول ورودی 138 توکن شد و توکن‌سازی 127 ثانیه طول کشید. نمونه‌ی اول برچسب 4 دارد و امتیاز واقعی‌اش 5 است، یعنی تبدیل درست انجام شده.</p>

```python
MAX_LEN = 384

class ReviewDS(torch.utils.data.Dataset):
    def __init__(self, d, with_labels=True):
        self.enc = tok(list(d['sm_l']), list(d['rt_l']), truncation='only_second', max_length=MAX_LEN)
        self.labels = (d['overall'].values - 1) if with_labels else None

    def __len__(self):
        return len(self.enc['input_ids'])

    def __getitem__(self, i):
        item = {k: self.enc[k][i] for k in self.enc.keys()}
        if self.labels is not None:
            item['labels'] = int(self.labels[i])
        return item

t0 = time.time()
train_ds = ReviewDS(tr)
val_ds = ReviewDS(va)
test_ds = ReviewDS(test, with_labels=False)
print(f'tokenize: {time.time() - t0:.0f} sanie | train {len(train_ds):,} | val {len(val_ds):,} | test {len(test_ds):,}')

lens_tr = np.array([len(x) for x in train_ds.enc['input_ids']])
print(f'miangin tool voroodi: {lens_tr.mean():.0f} token | bishine: {lens_tr.max()}')
print('kelid-haye har nemune:', list(train_ds[0].keys()))
print('label nemune aval:', train_ds[0]['labels'], '| emtiaz vaghei:', tr['overall'].iloc[0])
```

<div dir="rtl" style="direction:rtl;text-align:right;font-family:'B Nazanin', BNazanin, Nazanin, Tahoma, sans-serif;font-size:19px;line-height:2;color:#8fd0ff;background:transparent;padding:4px 56px 4px 8px">
<p dir="rtl" style="margin:0px 0 2px 0;font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:23px;font-weight:bold;color:#5ebcff;line-height:1.7">قدم 4. مدل و تنظیمات آموزش</p>
<p dir="rtl" style="margin:2px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">مسئله:</span> مدلی لازم بود که روی T4 رایگان در وقت معقول آموزش ببیند و زبان نظرهای کاربران را بشناسد.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">توضیح:</span> DistilRoBERTa (<span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;background:transparent;padding:0;direction:ltr;unicode-bidi:isolate">distilbert/distilroberta-base</span>) نسخه‌ی 6 لایه‌ای RoBERTa است، با 82 میلیون پارامتر، و تقریباً دو برابر سریع‌تر از نسخه‌ی کامل. RoBERTa روی حجم زیادی متن وب پیش‌آموزش دیده که به زبان نظرها نزدیک است. نرخ یادگیری 2e-5، بچ 32، دو ایپاک و وارم‌آپ شش درصدی تنظیمات رایج فاین‌تیون‌اند. بعد از هر ایپاک ارزیابی انجام می‌شود و بهترین مدل بر اساس micro-F1 نگه داشته می‌شود.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">راه حل:</span> <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;background:transparent;padding:0;direction:ltr;unicode-bidi:isolate">train_sampling_strategy='group_by_length'</span> نظرهای هم‌طول را در یک بچ می‌گذارد تا پدینگ و زمان کم شود. اول <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;background:transparent;padding:0;direction:ltr;unicode-bidi:isolate">group_by_length=True</span> نوشته بودیم و خط <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;background:transparent;padding:0;direction:ltr;unicode-bidi:isolate">parametr-haye nashenakhte</span> نشان داد این نسخه آن را نمی‌شناسد، چون اسمش در transformers 5 عوض شده. بعد از اصلاح این لیست خالی شد. جدول UNEXPECTED و MISSING طبیعی است: بخش پیش‌بینی کلمه (<span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;background:transparent;padding:0;direction:ltr;unicode-bidi:isolate">lm_head</span>) کنار گذاشته شد و لایه‌ی طبقه‌بندی تازه ساخته شد، که همان را آموزش می‌دهیم.</p>

```python
from transformers import (AutoModelForSequenceClassification, DataCollatorWithPadding,
                          Trainer, TrainingArguments, set_seed)
set_seed(SEED)

id2label = {i: str(i + 1) for i in range(5)}
model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME, num_labels=5, id2label=id2label, label2id={v: k for k, v in id2label.items()},
    dtype=torch.float32)

def compute_metrics(p):
    logits = p.predictions[0] if isinstance(p.predictions, tuple) else p.predictions
    pred = np.argmax(logits, axis=-1) + 1
    true = np.asarray(p.label_ids) + 1
    return {'micro_f1': f1_score(true, pred, average='micro'),
            'macro_f1': f1_score(true, pred, average='macro'),
            'within_1': float(np.mean(np.abs(true - pred) <= 1))}

RUN_NAME = f'distilroberta_n{len(tr)}_len{MAX_LEN}'
OUT_DIR = f'{BASE}/ckpt_{RUN_NAME}'
BATCH, EPOCHS, LR = 32, 2, 2e-5
steps_per_epoch = int(np.ceil(len(train_ds) / BATCH))

wanted = dict(
    output_dir=OUT_DIR, learning_rate=LR, num_train_epochs=EPOCHS,
    per_device_train_batch_size=BATCH, per_device_eval_batch_size=128,
    warmup_steps=int(0.06 * steps_per_epoch * EPOCHS), weight_decay=0.01,
    eval_strategy='epoch', save_strategy='epoch', load_best_model_at_end=True,
    metric_for_best_model='micro_f1', greater_is_better=True, save_total_limit=2,
    logging_steps=200, fp16=True, train_sampling_strategy='group_by_length' , dataloader_num_workers=2,
    report_to='none', seed=SEED)
allowed = inspect.signature(TrainingArguments.__init__).parameters
if 'eval_strategy' not in allowed:
    wanted['evaluation_strategy'] = wanted.pop('eval_strategy')
args = TrainingArguments(**{k: v for k, v in wanted.items() if k in allowed})
print('parametr-haye nashenakhte (hazf shodand):', [k for k in wanted if k not in allowed])

trainer = Trainer(model=model, args=args, train_dataset=train_ds, eval_dataset=val_ds,
                  data_collator=DataCollatorWithPadding(tok, pad_to_multiple_of=8),
                  compute_metrics=compute_metrics)
n_params = sum(p.numel() for p in model.parameters())
print(f'model: {MODEL_NAME} | parametr-ha: {n_params / 1e6:.0f}M | gam dar har epoch: {steps_per_epoch:,} | kol gam-ha: {steps_per_epoch * EPOCHS:,}')
```

<p dir="rtl" style="margin:2px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">مسئله:</span> کولب ممکن است وسط آموزش قطع شود و نزدیک نیم ساعت کار از بین برود.</p>

```python
import glob, re

def last_ckpt(d):
    cks = [p for p in glob.glob(f'{d}/checkpoint-*')
           if re.search(r'checkpoint-\d+$', p) and os.path.exists(f'{p}/trainer_state.json')]
    return max(cks, key=lambda p: int(p.rsplit('-', 1)[1])) if cks else None

last = last_ckpt(OUT_DIR) if os.path.isdir(OUT_DIR) else None
print('edame az:', last if last else 'shoroo az aval')

t0 = time.time()
trainer.train(resume_from_checkpoint=last)
print(f'zaman amoozesh: {(time.time() - t0) / 60:.1f} daghighe')

print('behtarin checkpoint:', trainer.state.best_model_checkpoint, '| micro_f1:', trainer.state.best_metric)
hist = pd.DataFrame([h for h in trainer.state.log_history if 'eval_micro_f1' in h])
cols = [c for c in ['epoch', 'eval_loss', 'eval_micro_f1', 'eval_macro_f1', 'eval_accuracy'] if c in hist.columns]
display(hist[cols].round(4))

BEST_DIR = f'{BASE}/best_{RUN_NAME}'
trainer.save_model(BEST_DIR)
tok.save_pretrained(BEST_DIR)
print('behtarin model zakhire shod dar:', BEST_DIR)
```

<p dir="rtl" style="margin:0px 0 2px 0;font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:23px;font-weight:bold;color:#5ebcff;line-height:1.7">قدم 6. پیش‌بینی احتمال‌ها با بهترین مدل</p>
<p dir="rtl" style="margin:2px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">مسئله:</span> اولین بار پیش‌بینی را با <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;background:transparent;padding:0;direction:ltr;unicode-bidi:isolate">trainer.predict</span> گرفتیم و دقت روی ولیدیشن معمولی 0.37 و روی ولیدیشن متوازن 0.20 شد، یعنی در حد شانس. خود Trainer موقع آموزش 0.7697 گزارش کرده بود.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">توضیح:</span> برچسب‌های داخل دیتاست با ترتیب ولیدیشن یکی بودند. پس مشکل از ترتیب خروجی <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;background:transparent;padding:0;direction:ltr;unicode-bidi:isolate">trainer.predict</span> بود، نه از مدل.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">راه حل:</span> پیش‌بینی با DataLoader خودمان و بدون shuffle انجام می‌شود. نظرها اول بر اساس طول مرتب می‌شوند تا پدینگ کم شود و بعد به ترتیب اصلی برمی‌گردند. اگر دقت روی ولیدیشن معمولی به 0.7697 نرسد، سلول متوقف می‌شود. خروجی با فایل‌هایی که قبلاً در درایو ذخیره شده هم مقایسه می‌شود، ولی روی آن‌ها نوشته نمی‌شود، چون نوت‌بوک اول در استکینگ از همان‌ها استفاده کرده.</p>

```python
import torch
from torch.utils.data import DataLoader, Subset
from transformers import AutoModelForSequenceClassification, DataCollatorWithPadding

RUN_NAME = f'distilroberta_n{len(tr)}_len{MAX_LEN}'
BEST_DIR = f'{BASE}/best_{RUN_NAME}'
device = 'cuda' if torch.cuda.is_available() else 'cpu'
best_model = AutoModelForSequenceClassification.from_pretrained(BEST_DIR, dtype=torch.float32).to(device).eval()
collator = DataCollatorWithPadding(tok, pad_to_multiple_of=8)

def softmax_np(z):
    z = z - z.max(axis=1, keepdims=True)
    e = np.exp(z)
    return e / e.sum(axis=1, keepdims=True)

def predict_proba(ds, bs=64):
    lens = np.array([len(ds[i]['input_ids']) for i in range(len(ds))])
    order = np.argsort(lens, kind='stable')
    dl = DataLoader(Subset(ds, order.tolist()), batch_size=bs, shuffle=False, collate_fn=collator)
    outs = []
    with torch.no_grad():
        for batch in dl:
            batch = {k: v.to(device) for k, v in batch.items() if k != 'labels'}
            with torch.autocast(device_type='cuda', dtype=torch.float16, enabled=(device == 'cuda')):
                outs.append(best_model(**batch).logits.float().cpu().numpy())
    P = np.concatenate(outs).astype(np.float64)
    out = np.empty_like(P)
    out[order] = P
    return softmax_np(out)

t0 = time.time()
val_proba_tr = predict_proba(val_ds)
test_proba_tr = predict_proba(test_ds)
y_va = va['overall'].values.astype(int)
acc_nat = accuracy_score(y_va, val_proba_tr.argmax(1) + 1)
print(f'zaman pishbini: {(time.time() - t0) / 60:.1f} daghighe | val: {val_proba_tr.shape} | test: {test_proba_tr.shape}')
print(f'deghat roye val tabiei: {acc_nat:.4f} (bayad 0.7697 bashad)')
assert acc_nat > 0.76, 'tartib ya model dorost nist'

for part, P in [('val', val_proba_tr), ('test', test_proba_tr)]:
    path = f'{BASE}/{part}_proba_{RUN_NAME}.npy'
    if os.path.exists(path):
        saved = np.load(path)
        print(f'{part}: hamkhani ba file zakhire-shode = {(saved.argmax(1) == P.argmax(1)).mean():.4f}')
    else:
        np.save(path, P)
        print(f'{part}: zakhire shod')
```

<p dir="rtl" style="margin:0px 0 2px 0;font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:23px;font-weight:bold;color:#5ebcff;line-height:1.7">قدم 7. اثر اصلاح پرایر و ترکیب ساده با TF-IDF</p>
<p dir="rtl" style="margin:2px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">مسئله:</span> ترنسفورمر هم روی ترینی آموزش دیده که 55 درصدش پنج ستاره است، پس همان اصلاح پرایر نوت‌بوک اول اینجا هم لازم است.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">توضیح:</span> α های مختلف امتحان می‌شوند و میانگین وزنی ترنسفورمر و TF-IDF هم سنجیده می‌شود. خط <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;background:transparent;padding:0;direction:ltr;unicode-bidi:isolate">barresi tartib</span> چک می‌کند که ولیدیشن متوازن و پرایر همان‌های نوت‌بوک اول باشند؛ عدد TF-IDF باید دقیقاً 0.5761 باشد.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">راه حل:</span> در اجرای اصلی، ترنسفورمر بدون اصلاح روی ولیدیشن متوازن 0.6464 گرفت و با <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;background:transparent;padding:0;direction:ltr;unicode-bidi:isolate">α=1</span> به 0.6626 رسید. بهترین ترکیب ساده 0.8 ترنسفورمر و 0.2 TF-IDF بود، با 0.6658. فایل <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;background:transparent;padding:0;direction:ltr;unicode-bidi:isolate">q2_submission_blend.csv</span> فقط پشتیبان است. ترکیب نهایی همراه با تاریخچه در قدم 25 نوت‌بوک اول به 0.6709 می‌رسد.</p>

```python
prior_tr = np.load(f'{BASE}/prior_train.npy')
bal_pos = np.load(f'{BASE}/val_balanced_pos.npy')
val_proba_tfidf = np.load(f'{BASE}/val_proba_tfidf.npy')
test_proba_tfidf = np.load(f'{BASE}/test_proba_tfidf.npy')
y_bal = y_va[bal_pos]
assert len(val_proba_tfidf) == len(va) == len(val_proba_tr), 'tedad radif-haye validation yeki nist'

def adj(P, alpha):
    p = P / prior_tr ** alpha
    return p / p.sum(axis=1, keepdims=True)

def bal_score(P, alpha=1.0):
    pred = np.argmax(adj(P, alpha), axis=1) + 1
    return f1_score(y_bal, pred[bal_pos], average='micro')

def make_submission(preds, path):
    preds = np.asarray(preds).astype(int).ravel()
    assert len(preds) == len(test_ds), 'tedad pishbini-ha ba test barabar nist'
    assert set(np.unique(preds)) <= {1, 2, 3, 4, 5}, 'label gheyr-mojaz'
    pd.DataFrame({'predicted': preds}).to_csv(path, index=False)
    check = pd.read_csv(path)
    assert list(check.columns) == ['predicted'] and len(check) == len(test_ds)
    print('zakhire shod:', path)
    print('tozi pishbini-ha:', check['predicted'].value_counts(normalize=True).sort_index().round(3).to_dict())

print('tedad har emtiaz dar val motavazen:', np.bincount(y_bal, minlength=6)[1:])
print(f'barresi tartib: TF-IDF roye val motavazen = {bal_score(val_proba_tfidf):.4f} (bayad 0.5761 bashad)')

rows = [{'model': 'TF-IDF + eslah prior', 'alpha': 1.0,
         'val_tabiei': accuracy_score(y_va, val_proba_tfidf.argmax(1) + 1), 'val_motavazen': bal_score(val_proba_tfidf)}]
for a in [0.0, 0.5, 0.75, 1.0, 1.25]:
    rows.append({'model': 'transformer', 'alpha': a,
                 'val_tabiei': accuracy_score(y_va, adj(val_proba_tr, a).argmax(1) + 1),
                 'val_motavazen': bal_score(val_proba_tr, a)})
display(pd.DataFrame(rows).round(4))

grid = pd.DataFrame({a: {round(w, 1): bal_score(w * val_proba_tr + (1 - w) * val_proba_tfidf, a)
                         for w in np.arange(0, 1.01, 0.1)} for a in [0.75, 1.0, 1.25]})
grid.index.name, grid.columns.name = 'w (vazn transformer)', 'alpha'
display(grid.round(4))
best_w, best_a = grid.stack().idxmax()
print(f'behtarin tarkib: w={best_w}, alpha={best_a} -> val motavazen = {grid.loc[best_w, best_a]:.4f}')

P_te = best_w * test_proba_tr + (1 - best_w) * test_proba_tfidf
make_submission(np.argmax(adj(P_te, best_a), axis=1) + 1, f'{BASE}/q2_submission_blend.csv')
```
