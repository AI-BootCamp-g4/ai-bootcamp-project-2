```python
def read_table(path):
    kw = dict(dtype={'asin': str, 'reviewerID': str}, low_memory=False)
    try:
        df = pd.read_csv(path, **kw)
    except UnicodeDecodeError:
        print(os.path.basename(path), 'ba utf-8 khande nashod; ba latin-1 khande mishavad')
        df = pd.read_csv(path, encoding='latin-1', **kw)
    df.columns = df.columns.astype(str).str.strip()
    extra = [c for c in df.columns if c.startswith('Unnamed')]
    if extra:
        print(os.path.basename(path), '| sotoon-e index-e ezafe hazf shod:', extra)
        df = df.drop(columns=extra)
    for c in ['asin', 'reviewerID']:
        if c in df.columns:
            df[c] = df[c].str.strip()
    return df

t0 = time.time()
train_raw = read_table(f'{BASE}/train.csv')
test = read_table(f'{BASE}/test.csv')
mapping = read_table(f'{BASE}/mapping.csv')
print(f'zaman-e khandan: {time.time() - t0:.0f} sanie')

for name, df in [('train', train_raw), ('test', test), ('mapping', mapping)]:
    print(f'{name}: {len(df):,} radif, {df.shape[1]} sotoon | hafeze: {df.memory_usage(deep=True).sum() / 1e9:.2f} GB')

t0 = time.time()
train_raw.to_pickle(f'{BASE}/train.pkl')
test.to_pickle(f'{BASE}/test.pkl')
mapping.to_pickle(f'{BASE}/mapping.pkl')
print(f'noskhe-ye sari (pkl) zakhire shod: {time.time() - t0:.0f} sanie')
```

```python
def text_report(df, name):
    rt = df['reviewText'].fillna('').astype(str)
    sm = df['summary'].fillna('').astype(str)
    e_rt, e_sm = rt.str.strip() == '', sm.str.strip() == ''
    print(f'===== {name}')
    print(f'reviewText-e khali: {e_rt.sum():,} ({e_rt.mean():.2%})')
    print(f'summary-e khali: {e_sm.sum():,} ({e_sm.mean():.2%})')
    print(f'har do khali: {(e_rt & e_sm).sum():,}')
    print(f'nazar-haye daraye HTML ya &quot;: {rt.str.contains(r"<[^>]+>|&[a-z]+;", regex=True).mean():.2%}')
    chars = rt.str.len()
    words = rt.str.split().str.len()
    print('tool-e matn (character) sadak 50/90/95/99:', chars.quantile([.5, .9, .95, .99]).round().astype(int).tolist(), '| bishine:', int(chars.max()))
    print('tool-e matn (kalame)    sadak 50/90/95/99:', words.quantile([.5, .9, .95, .99]).round().astype(int).tolist())
    print()

text_report(train_raw, 'train')
text_report(test, 'test')

chars_tr = train_raw['reviewText'].fillna('').astype(str).str.len()
print('miane-ye tool-e matn (character) baraye har emtiaz:')
display(chars_tr.groupby(y).median().astype(int).rename('median_chars'))
```

<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">راه حل:</span> 31,302 ردیف حذف شد و 807,642 ماند. تقسیم 90 به 10 طوری انجام شد که سهم هر امتیاز در دو بخش یکی بماند: 726,877 نظر ترین و 80,765 ولیدیشن، با سهم‌هایی که تا سه رقم اعشار یکی‌اند. خروجی در <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;background:transparent;padding:0;direction:ltr;unicode-bidi:isolate">q2_clean_split.pkl</span> ذخیره شد و نوت‌بوک ترنسفورمر هم همین فایل را می‌خواند.</p>

<p dir="rtl" style="margin:0px 0 2px 0;font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:23px;font-weight:bold;color:#5ebcff;line-height:1.7">قدم 12. پاک‌سازی متن برای TF-IDF</p>
<p dir="rtl" style="margin:2px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">مسئله:</span> پاک‌سازی معمولی کلمه‌هایی مثل not و but را هم حذف می‌کند. ولی «not good» دقیقاً برعکس «good» است.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">توضیح:</span> متن کوچک می‌شود، HTML و لینک پاک می‌شوند و <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;background:transparent;padding:0;direction:ltr;unicode-bidi:isolate">n't</span> به <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;background:transparent;padding:0;direction:ltr;unicode-bidi:isolate">not</span> تبدیل می‌شود. علامت‌های ! و ? می‌مانند، چون در نظرهای خیلی عصبانی یا خیلی راضی زیاد می‌آیند. 300 هزار نظر از ترین نمونه‌گیری شد، چون هر آموزش LogisticRegression روی همین حجم 5 تا 9 دقیقه طول کشید.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">راه حل:</span> کلمه‌های نفی و تضاد از لیست استاپ‌وردهای sklearn بیرون آمدند، مثل not، no، but، however، very و too. صورت پروژه هم هشدار داده بود که کلمه‌های مهم حذف نشوند. تست آخر خروجی نشان می‌دهد «I don't like it, it isn't good» به <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;background:transparent;padding:0;direction:ltr;unicode-bidi:isolate">i do not like it it is not good</span> تبدیل شد.</p>
</div>

```python
import re, html
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

NEG_KEEP = {'no', 'not', 'nor', 'never', 'none', 'nothing', 'nobody', 'neither', 'nowhere', 'cannot',
            'but', 'however', 'although', 'very', 'too', 'only', 'few', 'less', 'more', 'most', 'against'}
STOP = sorted(set(ENGLISH_STOP_WORDS) - NEG_KEEP)

TAG = re.compile(r'<[^>]+>')
URL = re.compile(r'https?://\S+|www\.\S+')
NT = re.compile(r"n['’]t\b")
NONWORD = re.compile(r'[^a-z0-9!? ]+')
SPACES = re.compile(r'\s+')

def clean(t):
    t = html.unescape(t)
    t = TAG.sub(' ', t)
    t = URL.sub(' ', t)
    t = t.lower()
    for a, b in [("can't", 'can not'), ("can’t", 'can not'), ("won't", 'will not'), ("won’t", 'will not')]:
        t = t.replace(a, b)
    t = NT.sub(' not', t)
    t = NONWORD.sub(' ', t)
    return SPACES.sub(' ', t).strip()

TFIDF_N = 300_000
tr_all = df[df['split'] == 'train']
if len(tr_all) > TFIDF_N:
    tr, _ = train_test_split(tr_all, train_size=TFIDF_N, stratify=tr_all['overall'], random_state=SEED)
else:
    tr = tr_all
tr = tr.copy()
va = df[df['split'] == 'val'].copy()
te = test.copy()

t0 = time.time()
for d in (tr, va, te):
    d['rt'] = d['reviewText'].map(clean)
    d['sm'] = d['summary'].map(clean)
print(f'paksazi: {time.time() - t0:.0f} sanie | train: {len(tr):,} | val: {len(va):,} | test: {len(te):,}')

print('\nsahm-e emtiaz-ha dar train-e nemune-giri-shode:', tr['overall'].value_counts(normalize=True).sort_index().round(4).tolist())
print('\nse nemune, ghabl va bad az paksazi:')
for _, r in tr.sample(3, random_state=SEED).iterrows():
    print('ghabl:', r['reviewText'][:160].replace('\n', ' '))
    print('bad:', r['rt'][:160])
    print()
print('azmoon-e kalamat-e nafy:', clean("I don't like it, it isn't good & it won't work <br/>!!"))
```

<div dir="rtl" style="direction:rtl;text-align:right;font-family:'B Nazanin', BNazanin, Nazanin, Tahoma, sans-serif;font-size:19px;line-height:2;color:#8fd0ff;background:transparent;padding:4px 56px 4px 8px">
<p dir="rtl" style="margin:0px 0 2px 0;font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:23px;font-weight:bold;color:#5ebcff;line-height:1.7">قدم 13. ساخت ویژگی‌های TF-IDF</p>
<p dir="rtl" style="margin:2px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">مسئله:</span> یک کلمه‌ی تنها همیشه کافی نیست. «good» و «not good» باید دو ویژگی جدا باشند.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">توضیح:</span> دو بردارساز جدا ساخته شد، یکی برای متن با 300 هزار ویژگی و یکی برای خلاصه با 50 هزار. کلمه‌های تکی و دوتایی هر دو گرفته شدند. <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;background:transparent;padding:0;direction:ltr;unicode-bidi:isolate">min_df=3</span> کلمه‌های خیلی نادر و غلط‌های تایپی را کنار می‌گذارد و <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;background:transparent;padding:0;direction:ltr;unicode-bidi:isolate">sublinear_tf</span> نمی‌گذارد تکرار یک کلمه در یک نظر بلند همه‌ی وزن را بگیرد.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">راه حل:</span> خلاصه جدا شد چون یک کلمه در خلاصه معمولاً مهم‌تر از همان کلمه وسط متن است. عبارت‌هایی مثل not good، stopped working و waste money در واژگان هستند. هر نظر به‌طور میانگین 86 ویژگی غیرصفر دارد و کل ماتریس ترین 0.21 گیگابایت جا گرفت.</p>
</div>

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from scipy.sparse import hstack
warnings.filterwarnings('ignore', message='.*stop_words may be inconsistent.*')

TOKEN = r'(?u)\b\w\w+\b|[!?]'
vec_rt = TfidfVectorizer(ngram_range=(1, 2), min_df=3, max_df=0.9, max_features=300_000,
                         sublinear_tf=True, stop_words=STOP, token_pattern=TOKEN, dtype=np.float32)
vec_sm = TfidfVectorizer(ngram_range=(1, 2), min_df=3, max_features=50_000,
                         sublinear_tf=True, token_pattern=TOKEN, dtype=np.float32)

t0 = time.time()
X_tr = hstack([vec_rt.fit_transform(tr['rt']), vec_sm.fit_transform(tr['sm'])]).tocsr()
X_va = hstack([vec_rt.transform(va['rt']), vec_sm.transform(va['sm'])]).tocsr()
X_te = hstack([vec_rt.transform(te['rt']), vec_sm.transform(te['sm'])]).tocsr()
y_tr, y_va = tr['overall'].values, va['overall'].values

print(f'zaman: {time.time() - t0:.0f} sanie')
print(f'andaze vazhegan: matn {len(vec_rt.vocabulary_):,} | kholase {len(vec_sm.vocabulary_):,}')
print('X_tr:', X_tr.shape, '| X_va:', X_va.shape, '| X_te:', X_te.shape)
print(f'miangin tedad vizhegi gheyr-sefr dar har nazar: {X_tr.getnnz(axis=1).mean():.0f}')
print(f'hafeze X_tr: {(X_tr.data.nbytes + X_tr.indices.nbytes + X_tr.indptr.nbytes) / 1e9:.2f} GB')
print('\njoft-kalame-haye manfi ke dar vazhegan hast:',
      [w for w in ['not good', 'not work', 'stopped working', 'waste money', 'not recommend', 'highly recommend']
       if w in vec_rt.vocabulary_])
```

<div dir="rtl" style="direction:rtl;text-align:right;font-family:'B Nazanin', BNazanin, Nazanin, Tahoma, sans-serif;font-size:19px;line-height:2;color:#8fd0ff;background:transparent;padding:4px 56px 4px 8px">
<p dir="rtl" style="margin:0px 0 2px 0;font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:23px;font-weight:bold;color:#5ebcff;line-height:1.7">قدم 14. خط پایه و LogisticRegression</p>
<p dir="rtl" style="margin:2px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">مسئله:</span> یک مدل مرجع ساده و سریع لازم بود تا بعداً ترنسفورمر با آن مقایسه شود.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">توضیح:</span> همیشه 5 گفتن micro-F1 برابر 0.5496 و macro-F1 برابر 0.1419 می‌دهد. LogisticRegression با <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;background:transparent;padding:0;direction:ltr;unicode-bidi:isolate">C=0.5</span> به 0.7076 رسید و با <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;background:transparent;padding:0;direction:ltr;unicode-bidi:isolate">C=1</span> به 0.7069، آن هم بعد از 550 ثانیه آموزش. C شدت منظم‌سازی را تعیین می‌کند.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">راه حل:</span> اجرای <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;background:transparent;padding:0;direction:ltr;unicode-bidi:isolate">C=2</span> و <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;background:transparent;padding:0;direction:ltr;unicode-bidi:isolate">C=4</span> را وسط کار قطع کردیم. هر کدام چند دقیقه طول می‌کشید و <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;background:transparent;padding:0;direction:ltr;unicode-bidi:isolate">C=1</span> از 0.5 بدتر شده بود، پس بزرگ‌تر کردن C امیدی نداشت. به‌جایش در قدم بعد <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;background:transparent;padding:0;direction:ltr;unicode-bidi:isolate">C=0.25</span> امتحان شد.</p>

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import f1_score, accuracy_score

def scores(y_true, y_pred):
    y_true, y_pred = np.asarray(y_true), np.asarray(y_pred)
    return {'micro_f1': f1_score(y_true, y_pred, average='micro'),
            'macro_f1': f1_score(y_true, y_pred, average='macro'),
            'accuracy': accuracy_score(y_true, y_pred),
            'within_1': float(np.mean(np.abs(y_true - y_pred) <= 1))}

results, runs = [], {}
base = scores(y_va, np.full(len(y_va), 5))
results.append({'model': 'baseline (always 5)', 'C': None, 'class_weight': None, **base, 'sec': 0})
print('baseline:', {k: round(v, 4) for k, v in base.items()})

def fit_eval(C, cw_name, cw):
    t0 = time.time()
    clf = LogisticRegression(C=C, class_weight=cw, max_iter=1000)
    clf.fit(X_tr, y_tr)
    pred = clf.predict(X_va)
    r = {'model': 'tfidf+LR', 'C': C, 'class_weight': cw_name, **scores(y_va, pred), 'sec': round(time.time() - t0)}
    runs[(cw_name, C)] = (clf, pred)
    results.append(r)
    print({k: (round(v, 4) if isinstance(v, float) else v) for k, v in r.items()})

for C in [0.5, 1]:
    fit_eval(C, 'none', None)
```

<p dir="rtl" style="margin:0px 0 2px 0;font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:23px;font-weight:bold;color:#5ebcff;line-height:1.7">قدم 15. امتحان <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;background:transparent;padding:0;direction:ltr;unicode-bidi:isolate">C=0.25</span> و انتخاب C</p>
<p dir="rtl" style="margin:2px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">مسئله:</span> شاید بهترین C کوچک‌تر از 0.5 باشد و ما فقط یک طرفش را دیده باشیم.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">توضیح:</span> <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;background:transparent;padding:0;direction:ltr;unicode-bidi:isolate">C=0.25</span> به 0.7038 رسید.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">راه حل:</span> این از 0.5 کمتر است. یعنی 0.5 بین دو مقدار بدتر قرار دارد، پس <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;background:transparent;padding:0;direction:ltr;unicode-bidi:isolate">C=0.5</span> انتخاب شد.</p>

```python
fit_eval(0.25, 'none', None)

best_C = max([C for (w, C) in runs if w == 'none'],
             key=lambda C: f1_score(y_va, runs[('none', C)][1], average='micro'))
print('\nbehtarin C:', best_C)
display(pd.DataFrame(results).round(4))
```

<p dir="rtl" style="margin:2px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">مسئله:</span> امتیاز 2 فقط 6.8 درصد داده است و مدل خیلی کم پیش‌بینی‌اش می‌کند.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">توضیح:</span> دو نوع وزن امتحان شد. وزن balanced به امتیاز 2 وزن 2.95 و به امتیاز 5 وزن 0.36 می‌دهد و وزن sqrt جذر همین است، یعنی ملایم‌تر. با sqrt، macro-F1 از 0.5384 به 0.5593 رفت ولی micro-F1 کمی پایین آمد (0.7048). با balanced، micro-F1 تا 0.6759 افتاد.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">راه حل:</span> روی ولیدیشن معمولی مدل بدون وزن بهتر بود. همین مقایسه در قدم 19 روی ولیدیشن متوازن تکرار شد و نتیجه برعکس شد.</p>

```python
from sklearn.metrics import confusion_matrix

pred_va_none = runs[('none', best_C)][1]
Cm = confusion_matrix(y_va, pred_va_none, labels=[1, 2, 3, 4, 5], normalize='true')
q_pred = np.bincount(test_proba_tfidf.argmax(axis=1), minlength=5) / len(test_proba_tfidf)
q_test = np.clip(np.linalg.solve(Cm.T, q_pred), 0, None)
q_test = q_test / q_test.sum()
prior_tr = np.bincount(y_tr - 1, minlength=5) / len(y_tr)
print('tozi emtiaz dar train         :', prior_tr.round(3))
print('tozi pishbini-ha roye test    :', q_pred.round(3))
print('takhmin tozi vaghei test      :', q_test.round(3))

rng = np.random.default_rng(SEED)
n_min = int(pd.Series(y_va).value_counts().min())
bal_pos = np.concatenate([rng.choice(np.flatnonzero(y_va == k), n_min, replace=False) for k in [1, 2, 3, 4, 5]])
y_bal = y_va[bal_pos]
print(f'\nValidation motavazen: {len(bal_pos):,} nazar ({n_min:,} az har emtiaz)')

def adj(proba, alpha):
    p = proba / prior_tr ** alpha
    return p / p.sum(axis=1, keepdims=True)

rows = []
for key in [('none', best_C), ('sqrt', best_C), ('balanced', best_C)]:
    pred = runs[key][1]
    rows.append({'model': f'class_weight={key[0]}',
                 'val_tabiei': f1_score(y_va, pred, average='micro'),
                 'val_motavazen': f1_score(y_bal, pred[bal_pos], average='micro')})
P_none = runs[('none', best_C)][0].predict_proba(X_va)
for a in [0.25, 0.5, 0.75, 1.0, 1.25]:
    pred = np.argmax(adj(P_none, a), axis=1) + 1
    rows.append({'model': f'none + eslah prior (alpha={a})',
                 'val_tabiei': f1_score(y_va, pred, average='micro'),
                 'val_motavazen': f1_score(y_bal, pred[bal_pos], average='micro')})
display(pd.DataFrame(rows).round(4))
```

```python
trn = df[df['split'] == 'train']
K5 = [1, 2, 3, 4, 5]
M = 3

rev_cnt = pd.crosstab(trn['reviewerID'], trn['overall']).reindex(columns=K5, fill_value=0)
prod_cnt = pd.crosstab(trn['asin'], trn['overall']).reindex(columns=K5, fill_value=0)

def hist_feats(d, cnt, key, prefix, subtract=None):
    C = cnt.reindex(d[key].values).fillna(0).to_numpy(dtype=float)
    if subtract is not None:
        C = np.clip(C - subtract, 0, None)
    n = C.sum(axis=1)
    dist = (C + M * prior_tr) / (n[:, None] + M)
    out = pd.DataFrame(dist, columns=[f'{prefix}_p{k}' for k in K5], index=d.index)
    out[f'{prefix}_n'] = n
    out[f'{prefix}_mean'] = dist @ np.array(K5, dtype=float)
    return out

def other_feats(d):
    ver = d['verified'].astype(str).str.strip().str.lower().isin(['true', '1', '1.0'])
    return pd.DataFrame({'verified': ver.astype(int).values,
                         'log_vote': np.log1p(d['vote_num'].values),
                         'len_text': d['reviewText'].str.len().values,
                         'len_summary': d['summary'].str.len().values}, index=d.index)

key_tr = (trn['reviewerID'] + '||' + trn['reviewText'].str.strip()).values
lab_by_key = pd.Series(trn['overall'].values, index=key_tr)
asin_by_key = pd.Series(trn['asin'].values, index=key_tr)
lab_by_key = lab_by_key[~lab_by_key.index.duplicated()]
asin_by_key = asin_by_key[~asin_by_key.index.duplicated()]
key_te = test['reviewerID'] + '||' + test['reviewText'].str.strip()
copy_lab = key_te.map(lab_by_key)
copy_asin = key_te.map(asin_by_key)
has_copy = copy_lab.notna().values
same_prod = has_copy & (copy_asin.values == test['asin'].values)
sub_rev = np.zeros((len(test), 5))
sub_prod = np.zeros((len(test), 5))
sub_rev[np.flatnonzero(has_copy), copy_lab[has_copy].astype(int).values - 1] = 1
sub_prod[np.flatnonzero(same_prod), copy_lab[same_prod].astype(int).values - 1] = 1
print(f'kopi-haye test dar train: {has_copy.sum():,} | az tarikhche nazardahande kam shod; '
      f'{same_prod.sum():,} ta az tarikhche mahsool ham')

F_va = pd.concat([hist_feats(va, rev_cnt, 'reviewerID', 'rev'),
                  hist_feats(va, prod_cnt, 'asin', 'prod'), other_feats(va)], axis=1)
F_te = pd.concat([hist_feats(test, rev_cnt, 'reviewerID', 'rev', sub_rev),
                  hist_feats(test, prod_cnt, 'asin', 'prod', sub_prod), other_feats(test)], axis=1)
F_va.to_pickle(f'{BASE}/F_va.pkl')
F_te.to_pickle(f'{BASE}/F_te.pkl')
print('F_va:', F_va.shape, '| F_te:', F_te.shape)
print(F_va[['rev_n', 'rev_mean', 'prod_n', 'prod_mean', 'log_vote', 'len_text']].describe().round(2).T[['mean', '50%', 'max']])
```

<p dir="rtl" style="margin:0px 0 2px 0;font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:23px;font-weight:bold;color:#5ebcff;line-height:1.7">قدم 23. استکینگ با تاریخچه و بدون آن</p>
<p dir="rtl" style="margin:2px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">مسئله:</span> احتمال‌های TF-IDF را چطور با ویژگی‌های تاریخچه ترکیب کنیم، بدون اینکه مدل دوم روی پیش‌بینی‌هایی آموزش ببیند که بیش از حد خوش‌بین‌اند؟</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">توضیح:</span> احتمال‌های TF-IDF روی خود ترین زیادی مطمئن‌اند، چون مدل آن نظرها را دیده. روی ولیدیشن ولی صادق‌اند. پس مدل دوم (HistGradientBoosting) روی ولیدیشن با 5 فولد آموزش می‌بیند و هر نظر پیش‌بینی‌ای می‌گیرد که موقع آموزش دیده نشده. مدل درختی انتخاب شد چون خودش قاعده‌ای مثل «اگر نظرهای قبلی زیاد است به تاریخچه بیشتر اعتماد کن» را پیدا می‌کند.</p>
<p dir="rtl" style="margin:6px 0 0 0;text-align:right"><span style="font-family:'B Titr', BTitr, Titr, 'B Nazanin', Tahoma, sans-serif;font-size:17px;font-weight:bold;color:#5ebcff">راه حل:</span> نسخه‌ی A فقط احتمال‌های TF-IDF را می‌گیرد و نسخه‌ی B تاریخچه را هم. روی ولیدیشن متوازن TF-IDF تنها 0.5761 گرفت، A برابر 0.5843 و B برابر 0.5948. یعنی تاریخچه حدود 1 درصد اضافه کرد. این بار هم <span dir="ltr" style="font-family:Consolas,Menlo,monospace;font-size:15px;color:#ffcf70;background:transparent;padding:0;direction:ltr;unicode-bidi:isolate">α=1</span> بهترین بود.</p>

```python
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.model_selection import StratifiedKFold

def stack_cv(X, y, X_test):
    oof = np.zeros((len(X), 5))
    test_p = np.zeros((len(X_test), 5))
    skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)
    for f, (a, b) in enumerate(skf.split(X, y)):
        m = HistGradientBoostingClassifier(learning_rate=0.05, max_iter=600, max_leaf_nodes=31,
                                           early_stopping=True, validation_fraction=0.1,
                                           n_iter_no_change=30, random_state=SEED)
        m.fit(X[a], y[a])
        oof[b] = m.predict_proba(X[b])
        test_p += m.predict_proba(X_test) / 5
        print(f'  fold {f + 1}: {m.n_iter_} derakht')
    return oof, test_p

def bal_score(P, alpha=1.0):
    pred = np.argmax(adj(P, alpha), axis=1) + 1
    return f1_score(y_bal, pred[bal_pos], average='micro')

X_tf_va, X_tf_te = val_proba_tfidf, test_proba_tfidf
X_all_va = np.hstack([val_proba_tfidf, F_va.to_numpy(dtype=float)])
X_all_te = np.hstack([test_proba_tfidf, F_te.to_numpy(dtype=float)])

t0 = time.time()
print('A) faghat ehtemal-haye TF-IDF:')
oof_A, te_A = stack_cv(X_tf_va, y_va, X_tf_te)
print('B) TF-IDF + tarikhche + sotoon-haye digar:')
oof_B, te_B = stack_cv(X_all_va, y_va, X_all_te)
print(f'zaman: {time.time() - t0:.0f} sanie\n')

rows = [{'model': 'TF-IDF + eslah prior (bedoon stacking)', 'val_motavazen': bal_score(val_proba_tfidf)}]
for name, P in [('stacking A: faghat TF-IDF', oof_A), ('stacking B: TF-IDF + tarikhche', oof_B)]:
    for a in [0.75, 1.0, 1.25]:
        rows.append({'model': f'{name} (alpha={a})', 'val_motavazen': bal_score(P, a)})
display(pd.DataFrame(rows).round(4))
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score

def pred_dist(P):
    return np.bincount(P.argmax(axis=1), minlength=5) / len(P)

print('tozi emtiaz dar train          :', prior_tr.round(3))
print('pishbini transformer roye val  :', pred_dist(val_proba_trf).round(3))
print('pishbini transformer roye test :', pred_dist(test_proba_trf).round(3))

Cm_trf = confusion_matrix(y_va, val_proba_trf.argmax(axis=1) + 1, labels=[1, 2, 3, 4, 5], normalize='true')
q_test_trf = np.clip(np.linalg.solve(Cm_trf.T, pred_dist(test_proba_trf)), 0, None)
q_test_trf = q_test_trf / q_test_trf.sum()
print('takhmin tozi vaghei test (BBSE):', q_test_trf.round(3))

scen_rows = []
for name, Pv in [('transformer', val_proba_trf), ('stacking C', oof_C)]:
    for a in [0.0, 1.0]:
        pred = np.argmax(adj(Pv, a), axis=1) + 1
        scen_rows.append({'model': name, 'eslah prior': 'bale' if a else 'na',
                          'agar test mesl train bashad': accuracy_score(y_va, pred),
                          'agar test motavazen bashad': accuracy_score(y_bal, pred[bal_pos])})
display(pd.DataFrame(scen_rows).round(4))

make_submission(np.argmax(te_C, axis=1) + 1, f'{BASE}/q2_submission_bedoon_eslah.csv')
```

```python
from sklearn.metrics import recall_score

share = np.bincount(y_va, minlength=6)[1:] / len(y_va)
P_best_val = cands[best['model']][0]
rec_models = {'transformer (bedoon eslah)': adj(val_proba_trf, 0.0),
              f"model nahayi (alpha={best['alpha']})": adj(P_best_val, best['alpha'])}

rec_rows = []
for name, P in rec_models.items():
    rec = recall_score(y_va, P.argmax(axis=1) + 1, labels=[1, 2, 3, 4, 5], average=None)
    rec_rows.append({'model': name, **{f'emtiaz {k}': r for k, r in zip(range(1, 6), rec)},
                     'miangin vazni (val tabiei)': (share * rec).sum(),
                     'miangin sade (val motavazen)': rec.mean()})
rec_table = pd.DataFrame(rec_rows).set_index('model')
print('sahm har emtiaz dar val tabiei:', share.round(3))
display(rec_table.round(3))

xk = np.arange(1, 6)
fig, ax = plt.subplots(figsize=(8, 4.6))
for i, (name, r) in enumerate(rec_table.iterrows()):
    ax.bar(xk + (i - 0.5) * 0.38, [r[f'emtiaz {k}'] for k in xk], width=0.38, label=name)
ax.set_xticks(xk)
ax.set_xticklabels([f'{k}\n({s:.0%} of val)' for k, s in zip(xk, share)])
ax.set_ylabel('recall (share predicted correctly)')
ax.set_ylim(0, 1)
ax.set_title('Same model, different weights: per-rating accuracy')
ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.22), ncol=2, frameon=False)
plt.tight_layout()
plt.savefig(f'{BASE}/recall_per_rating.png', dpi=150)
plt.show()
```

```python
from sklearn.metrics import f1_score, recall_score

metric_models = {'TF-IDF': val_proba_tfidf, 'transformer': val_proba_trf,
                 'stacking B (TF-IDF + tarikhche)': oof_B, 'stacking C (nahayi)': oof_C}
metric_rows = []
for name, Pv in metric_models.items():
    for a in [0.0, 1.0]:
        pred = np.argmax(adj(Pv, a), axis=1) + 1
        metric_rows.append({'model': name, 'eslah prior': 'bale' if a else 'na',
                            'micro-F1 (val tabiei)': f1_score(y_va, pred, average='micro'),
                            'macro-F1 (val tabiei)': f1_score(y_va, pred, average='macro'),
                            'miangin recall (val tabiei)': recall_score(y_va, pred, average='macro'),
                            'micro-F1 (val motavazen)': f1_score(y_bal, pred[bal_pos], average='micro')})
metrics_table = pd.DataFrame(metric_rows)
metrics_table.to_csv(f'{BASE}/results_final.csv', index=False)
pd.set_option('display.width', 250)
display(metrics_table.round(4))
print('zakhire shod: results_final.csv')
```
